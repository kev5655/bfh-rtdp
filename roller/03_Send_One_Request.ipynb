# Send one Request
- This notebook shows, how to send one request to a server.
- The Flask-Server must be running in order to respond to the request

## Prerequisite in the server-side script
<pre>
@server.route('/generator')
def generator():
    data = json.loads(request.data)
    print(data)
    return(data)
</pre>

```python
import requests as rq
ip = 'http://your-ip:5000'
```

```python
# send one request and display the response
data = '{"hello": "world", "success": "yes"}'
response = rq.get(ip+"/generator", json=data)
print(response.content)
```
